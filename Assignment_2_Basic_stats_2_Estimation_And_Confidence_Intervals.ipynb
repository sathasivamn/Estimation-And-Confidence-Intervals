# Estimation And Confidence Intervals

```python
# Import libraries
import numpy as np
import scipy.stats as stats
```

```python
# Given durability data (in millions of characters)
durability = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
# Sample Size
n = len(durability)
```

```python
# Sample Mean
sample_mean = np.mean(durability)
```

```python
# Sample Standard Deviation (used when population SD is unknown)
## Delta Degrees of Freedom; setting ddof=1 ensures that the sample standard deviation is calculated (instead of population SD).
sample_std = np.std(durability, ddof=1)  # ddof=1 for sample standard deviation
```

```python
# Confidence Level
confidence_level = 0.99
alpha = 1 - confidence_level
```

## A. 99% Confidence Interval using Sample Standard Deviation

```python
# Since population SD is unknown, use t-distribution
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)  # t-value for 99% confidence
```

```python
# Margin of Error
margin_error_t = t_critical * (sample_std / np.sqrt(n))
```

```python
# Confidence Interval using t-distribution
ci_lower_t = sample_mean - margin_error_t
ci_upper_t = sample_mean + margin_error_t
```

```python
print(f" 99% Confidence Interval (Using Sample SD): ({ci_lower_t:.4f}, {ci_upper_t:.4f})")
```

# B. 99% Confidence Interval using Known Population Standard Deviation

```python
# Given Population Standard Deviation
pop_std = 0.2
```

```python
# # Since population SD is known, use normal distribution (z-distribution)
z_critical = stats.norm.ppf(1 - alpha/2)  # z-value for 99% confidence
```

```python
# Margin of Error
margin_error_z = z_critical * (pop_std / np.sqrt(n))
```

```python
# Confidence Interval using z-distribution
ci_lower_z = sample_mean - margin_error_z
ci_upper_z = sample_mean + margin_error_z
```

```python
print(f" 99% Confidence Interval (Using Population SD = 0.2): ({ci_lower_z:.4f}, {ci_upper_z:.4f})")
```

# Conclusion

## Result
### The confidence interval width will be larger when using the t-distribution because of increased uncertainty due to estimating standard deviation.

### The z-distribution CI is narrower, as we assume knowledge of the exact population standard deviation.
